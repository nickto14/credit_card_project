Now we can check for duplicate values. We dont see any duplicates, but there are duplicates for the transaction ID variable. This variable can be dropped, as it will not add any information to our models.

```python
print(df.duplicated().sum())
print(df['Transaction ID'].duplicated().sum())
```

After applying the transformations and changes to our data, we can now take a look at the cleaned dataset.

Now lets create some visuals to investigate the data further.

1. We plot transactions on the y-axis and time on the x-axis 
- The majority of transactions occur between 7am and 11pm.

```python
fig = px.histogram(df_clean, x="Time")
fig.show()
```

2. We now do a scatterplot of transaction amount:
- We see most fraud occurs for purchases < $50. However, we also see a jump in fraudulent transactions around the $350-$400 price range. 

```python
df_clean['Fraud'] = df_clean['Fraud'].astype(str)
fig = px.histogram(df_clean, x="Amount", color="Fraud")
fig.show()
```
